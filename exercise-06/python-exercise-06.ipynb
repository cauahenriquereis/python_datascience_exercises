```python
import pandas as pd

df = pd.read_csv('paises.csv', sep=';', encoding='utf-8-sig')

df['Country'] = df['Country'].str.strip()
df['Region'] = df['Region'].str.strip()

df.head()
```

Questão 1

```python
oceania = df[df['Region'].str.contains('OCEANIA')]

print('Países da Oceania:')
print(oceania['Country'].to_string(index=False))
print('Quantidade de países da Oceania:', len(oceania))
```

Questão 2

```python
maior = df.loc[df['Population'].idxmax()]

print('País:', maior['Country'])
print('Região:', maior['Region'])
print('População:', maior['Population'])
```

Questão 3

```python
media_alfabetizacao = df.groupby('Region')['Literacy (%)'].mean().sort_values(ascending=False)
media_alfabetizacao
```

Questão 4

```python
sem_costa = df[df['Coastline (coast/area ratio)'] == 0]['Country']

sem_costa.to_csv('noCoast.csv', index=False)

print('Quantidade de países sem costa:', len(sem_costa))
sem_costa
```

Questão 5

```python
def classificar_mortalidade(deathrate):
    if deathrate < 9:
        return 'Balanced'
    return 'Urgent'

df['Humanitarian Help'] = df['Deathrate'].apply(classificar_mortalidade)
df
```
