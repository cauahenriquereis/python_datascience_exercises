Exercício 1

```python
import numpy as np

dataset = np.loadtxt('space.csv', delimiter = ';', dtype = str, encoding= 'utf-8')

header = dataset[0]

body = dataset[1:]

success = body[:,7] == 'Success'
failure =  body[:,7] == 'Failure'

total = np.sum(success) + np.sum(failure)

print(f"Porcentagem de missões que deram certo: {np.sum(success)/total * 100: .2f}%")
```

**Exercício 2**

```python
values = body[:, 6].astype(float)

values_available = values[values > 0]

media = np.mean(values_available)

print(f'Média de gastos {media: .2f}')

```
