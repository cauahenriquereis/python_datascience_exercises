```python
import pandas as pd
import numpy as np
```

Questão 1

```python
seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})

print(seriesAno1)
print(seriesAno2)
```

Questão 2

```python
totalAno1 = seriesAno1.sum()
totalAno2 = seriesAno2.sum()

print(f"Total ano 1: {totalAno1: .2f}%")
print(f"Total ano 2: {totalAno2: .2f}%")
```

Questão 3

```python
crescimento = seriesAno2 - seriesAno1

print(crescimento)
```

Questão 4

```python
crescimentoPositivo = crescimento[crescimento > 0]

print(crescimentoPositivo)
```

Questão 5

```python
projecao = seriesAno2 + crescimento * 2

maisPopular = projecao.nlargest(1)

print(maisPopular)
```

Questão 6

```python
df = pd.DataFrame(
    index = ['A', 'B', 'C', 'D', 'E'],
    columns = ['W', 'X', 'Y', 'Z'],
    data = np.random.randint(1, 50, [5, 4])
)

print(df)

mediaColunaX = df['X'][df['X'] < 30].mean()

print(f"Média da coluna X (menores que 30): {mediaColunaX: .2f}")
```

Questão 7

```python
mediaLinhaD = df.loc['D'].mean()
somaLinhaE = df.iloc[4].sum()

print(f"Média da linha D: {mediaLinhaD: .2f}")
print(f"Soma da linha E: {somaLinhaE}")
```

Questão 8

```python
sliceDf = df.loc[['A', 'C', 'E'], ['X', 'Y']]

print(sliceDf)

somaLinhas = sliceDf.sum(axis=1)
somaColunas = sliceDf.sum(axis=0)

print(somaLinhas)
print(somaColunas)
```
